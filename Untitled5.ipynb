```python
%%R

install.packages("BiocManager")

BiocManager::install(c(
    "ensembldb",
    "EnsDb.Hsapiens.v86",
    "GenomicFeatures",
    "AnnotationFilter"
))
```

```python
%%R

library(ensembldb)
library(EnsDb.Hsapiens.v86)
library(AnnotationFilter)
library(GenomicFeatures)

# Load Ensembl human database
edb <- EnsDb.Hsapiens.v86
```

```python
%%R

gene_name <- "ACTC1"

gene_data <- genes(edb,
                   filter = GeneNameFilter(gene_name))

gene_data
```

```python
%%R

tx <- transcripts(
    edb,
    filter = GeneNameFilter("ACTC1"),
    return.type = "DataFrame"
)

tx
```

```python
%%R

write.csv(tx,
          file = "ACTC1_transcripts.csv",
          row.names = FALSE)
```

```python
%%R

prot <- proteins(
    edb,
    filter = GeneNameFilter("ACTC1")
)

prot
```

```python
%%R

write.csv(as.data.frame(prot),
          file = "ACTC1_proteins.csv",
          row.names = FALSE)
```

```python
%%R

protein_ids <- names(prot)

protein_ids
```

```python
%%R

library(IRanges)

# Example protein ID
prot_id <- protein_ids[1]

# Create IRanges object
ir <- IRanges(start = 1, end = 100)

# Assign protein ID as names
names(ir) <- prot_id

# Map protein coordinates to genome
protein_map <- proteinToGenome(
    ir,
    edb
)

protein_map
```

```python
%%R

saveRDS(protein_map,
        file = "ACTC1_protein_genome_map.rds")
```
